Chai Bill Calculator in python:


```python
def chai_bill(cups, *args, **kwargs):
    total_cups = cups * 15
    total_add_ons = 0
    
    for item in args:
        if item in ("ginger", "masala", "biscuits"):
            total_add_ons += 5
            
    total = total_cups + total_add_ons 
    
    discount_percent = kwargs.get('discount_percent', 0)
    
    return total - (total * discount_percent / 100)

```

```python
print(chai_bill(2))
print(chai_bill(3, 'ginger', 'masala'))
print(chai_bill(4, 'biscuits', discount_percent = 10))
print(chai_bill(5, 'ginger', 'masala', 'biscuits', discount_percent = 20))
```

Youtub channel class implementation (Assignment):


```python
class YouTubeChannel:
    def __init__(self, name):
        self.name = name
        self.count = 0
    
    def subscribe(self):
        self.count += 1
        print(f"Thank you for subscribing: Total:{self.count}")
    
    def unsubscribe(self):
        if self.count > 0:
            self.count -= 1
            print(f"Total: {self.count}")

    
    def show(self):
        print(f"{self.name} has {self.count} subscribers")
```

```python
class chai_bill:
    def __init__(self, cups: int, price_per_cup: int) -> None:
        self.cups = cups
        self.price_per_cup = price_per_cup 
        
    def total_price(self) -> int:
        try:
            if self.cups > 0:
                return self.cups * self.price_per_cup
        
        except Exception as e:
            return {e}
    def show(self) -> str:
        return f"{self.cups} cups x ₹{self.price_per_cup} = ₹{self.total_price()}"
```

Now, adding memory to the chatbot:


```python
conversation = []

def chat(user_message):
    conversation.append({"role":"user", "parts":[{"text":user_message}]})
    
    try:
        client = genai.Client()
        response = client.models.generate_content(
            model = "gemini-3.5-flash", 
            contents = conversation
        )
        
        reply = response.text
    except Exception as e:
        return f"sorry! Something went wrong !{e}"
    
    conversation.append({
        "role":"model", 
        "parts":[{"text":reply}]
    }) 
    
    return reply
```

Now making the LLM create a recipe with a structured output:


```python
from pydantic import BaseModel
from google.genai import types
from google import genai
from dotenv import load_dotenv

load_dotenv()
conversations = []

class recipe(BaseModel):
    title: str
    ingredients: list[str]
    calories: int
    prep_time_minutes: int

config = types.GenerateContentConfig(
    response_mime_type = "application/json",
    response_schema = recipe, 
)

def get_recipe(user_input):
    conversations.append({
        "role":"user", 
        "parts": [{"text": user_input}]
    })
    
    client = genai.Client()
    
    response = client.models.generate_content(
        model = "gemini-3.5-flash", 
        contents = conversations, 
        config = config
    )
    reply = response.text
    
    conversations.append({
        "role":"model", 
        "parts":[{"text":reply}]
    })
    
    return response.parsed
    
```

Customer Review Analyzer: (Exercise)


```python
from google import genai
from google.genai import types
from dotenv import load_dotenv
from pydantic import BaseModel

load_dotenv()
conversations = []

class ReviewAnalyzer(BaseModel):
    sentiment: str
    ratings: int
    pros: list[str]
    cons: list[str]

config = types.GenerateContentConfig(
    response_mime_type = "application/json", 
    response_schema = ReviewAnalyzer, 
)
def get_customer_review_analysis(user_input):
    conversations.append({
        "role":"user", 
        "parts": [{"text":user_input}]
    })
    
    client = genai.Client()
    
    responses = client.models.generate_content(
        model = "gemini-3.5-flash", 
        contents = conversations, 
        config = config
    )
    
    reply = responses.text
    
    conversations.append({
        "role": "model", 
        "parts": [{"text": reply}]
    })
    
    return responses.parsed
```

```python
result = get_customer_review_analysis("I ordered the wireless earbuds last week. Sound quality is amazing and battery lasts all day, but the case feels cheap and it arrived two days late. Probably, I won't buy from this brand again.")
```

Complete Document AI Summarizer:


```python
from google import genai
from google.genai import types
from pydantic import BaseModel
from dotenv import load_dotenv

load_dotenv()

conversations = []

with open("deep_learning_trends.txt", "r", encoding = "utf-8") as f:
    document = f.read()


class summerizer(BaseModel):
    title: str
    summary: list[str]
    topics: list[str]

config = types.GenerateContentConfig(
    response_mime_type = "application/json", 
    response_schema = summerizer, 
)

def summary(user_input, document):
    conversations.append({
        "role":"user", 
        "parts":[{"text": user_input}, 
                 {"text":document}]
    })
    
    client = genai.Client()
    
    response = client.models.generate_content(
        model = "gemini-3.5-flash", 
        contents = conversations, 
        config = config
    )
    
    reply = response.text
    
    conversations.append({
        "role":"model", 
        "parts":[{"text":reply}]
    })
    
    with open("Summary.txt", "w", encoding="utf-8") as f:
        f.write(response.parsed.model_dump_json(indent = 2))
    
    return response.parsed
```

```python
result = summary("Summerize this document", document)
```

```python
result.summary
```

```python
for topic in result.topics:
    print(topic)
```

```python
with open("Summary.txt", "r", encoding = "utf-8") as f:
    content = f.read()
    print(content)
```
